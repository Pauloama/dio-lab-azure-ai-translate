```python
import requests
from bs4 import BeautifulSoup

def extract_text_from_url(url):
    response = requests.get(url)

    if response.status_code == 200:
        soup = BeautifulSoup(response.content, 'html.parser')
        for script in soup(["script", "style"]):
            script.decompose()
        text = soup.get_text(separator=' ')
        lines = (line.strip() for line in text.splitlines())
        chunks = (phrase.strip() for line in lines for phrase in line.split("  "))
        return " ".join(chunks)
    else:
        print(f"Failed to retrieve the page. Status code: {response.status_code}")
        return None
        
    return soup.get_text()
extract_text_from_url('https://dev.to/dannwaneri/the-token-economy-3cd9')
```

```python
from langchain_openai.chat_models.azure import AzureChatOpenAI

client = AzureChatOpenAI(
    azure_endpoint="https://dio-paulo-bootcamp-dev-eastus.openai.azure.com/",
    api_key="",
    api_version="2025-11-13",
    deployment_name="gpt-5.1-codex-mini",
    max_retries=0
    )

def translate_article(text, lang):
    messages = [
        ("system", "You are a helpful assistant that translates articles."),
        ("user", f"Translate the following article to {lang}:\n\n{text}")
    ]

    response = client.invoke(messages)
    print(response.content)
    return response.content

translate_article('https://dev.to/dannwaneri/the-token-economy-3cd9', 'Portuguese')
```

```python
url = 'https://dev.to/dannwaneri/the-token-economy-3cd9'
text = extract_text_from_url(url)
article = translate_article(text, 'Portuguese')

print(article)
```
